```python
import pandas as pd
from pathlib import Path
from property_prediction.dataset import get_db_connection
from dotenv import dotenv_values
from matminer.featurizers.conversions import StrToComposition
from matminer.featurizers.composition import ElementProperty

pd.set_option('display.max_columns', 200)

```

```python
ROOT = Path().resolve().parent
config = dotenv_values(ROOT / '.env')
conn = get_db_connection(config)
db_name = 'materials'

sql = f"SELECT * from {db_name}"
raw_materials_df = pd.read_sql(sql, conn)
```

```python
formulas_df = raw_materials_df[['formula_pretty']]
formulas_df = formulas_df.drop_duplicates(subset='formula_pretty')
str_to_comp = StrToComposition()
str_to_comp.set_n_jobs(8)
```

```python
processed_df = str_to_comp.featurize_dataframe(formulas_df, 'formula_pretty')
```

```python
ep = ElementProperty.from_preset(preset_name='magpie')
ep.set_n_jobs(8)
comp = processed_df.iloc[0,1]
features_df = ep.featurize_dataframe(processed_df, 'composition')
```

```python
merged_df = raw_materials_df[['material_id', 'formula_pretty']].merge(features_df, on='formula_pretty').drop(columns=['formula_pretty'])
```

```python
sql = """
    SELECT material_id, property_name, value
    from properties
    WHERE property_name IN ('band_gap', 'formation_energy_per_atom', 'bulk_modulus')
    AND value IS NOT NULL
"""
properties_df = pd.read_sql(sql, conn)
```

```python
properties_pivot = properties_df.pivot_table(index='material_id', columns='property_name', values='value').reset_index()
final_df = merged_df.merge(properties_pivot, on='material_id', how='inner')
final_df.to_parquet('../data/processed/final_data.parquet')
```

```python
conn.close()
```
